```python
# STUDENT INFORMATION SETUP

last_name = "dela Cruz"
seed_num = 9                  
favorite_artist = "Shirebound and Busking"     
```

```python
# EXERCISE 1: Engineering Test Data Organizer

sources = [f"{last_name.replace(' ', '')}_T{i}" for i in range(1, 6)]
categories = ["TEMPERATURE", "VOLTAGE", "PRESSURE", "SPEED"]

test_records = [
    {"source": sources[0], "category": categories[0], "value": seed_num * len(favorite_artist)},
    {"source": sources[1], "category": categories[1], "value": seed_num + 12.0},
    {"source": sources[2], "category": categories[2], "value": seed_num * 5.1},
    {"source": sources[3], "category": categories[3], "value": seed_num + float(favorite_artist.find('a'))},
    {"source": sources[4], "category": categories[0], "value": seed_num * 3.2},
    {"source": sources[0], "category": categories[0], "value": seed_num * len(favorite_artist)}  
]

unique_records = [dict(t) for t in {tuple(d.items()) for d in test_records}]

print("=== EXERCISE 1: GENERATED TEST RECORDS ===")
for r in test_records:
    print(r)

print("\n=== EXERCISE 1: ORGANIZED & DE-DUPLICATED DATA ===")
for r in unique_records:
    print(r)
```

```python
# EXERCISE 2: Student Performance Data Analyzer

from collections import defaultdict

students = [f"{last_name.replace(' ', '')}_S{i}" for i in range(1, 6)]
subjects = ["PROGRAMMING", "MATHEMATICS", "PHYSICS", "ELECTRONICS", "ENGLISH"]

performance_records = [
    {"student": students[0], "subject": subjects[0], "score": seed_num * 10 + 2.5},
    {"student": students[1], "subject": subjects[1], "score": 74.0},
    {"student": students[2], "subject": subjects[2], "score": 92.0},
    {"student": students[3], "subject": subjects[3], "score": None},  
    {"student": students[4], "subject": subjects[4], "score": 81.5},
    {"student": students[0], "subject": subjects[0], "score": seed_num * 10 + 2.5}   
]

subject_summaries = defaultdict(list)
for rec in performance_records:
    if rec["score"] is not None:
        subject_summaries[rec["subject"]].append(rec["score"])

print("\n=== EXERCISE 2: STUDENT PERFORMANCE REPORT ===")
for subj, scores in subject_summaries.items():
    avg = sum(scores) / len(scores)
    print(f"Subject: {subj} | Scores: {scores} | Average: {avg:.2f}")
```

```python
# EXERCISE 3: Transaction History Analyzer

customers = [f"{last_name.replace(' ', '')}_C{i}" for i in range(1, 6)]
product_prices = {
    "NOTEBOOK": 50.0,
    "PEN": 15.0,
    "USB": 350.0,
    "HEADPHONES": 1200.0,
    "CALCULATOR": 850.0
}

transactions = [
    {"customer": customers[0], "product": "NOTEBOOK", "qty": seed_num},
    {"customer": customers[1], "product": "USB", "qty": 2},
    {"customer": customers[2], "product": "PEN", "qty": seed_num + 5},
    {"customer": customers[3], "product": "HEADPHONES", "qty": 1},
    {"customer": customers[0], "product": "NOTEBOOK", "qty": seed_num}  # Repeated Purchase
]

grand_total = 0
print("\n=== EXERCISE 3: CONSOLIDATED TRANSACTION REPORT ===")
for tx in transactions:
    line_total = tx["qty"] * product_prices[tx["product"]]
    grand_total += line_total
    print(f"Customer: {tx['customer']} | Product: {tx['product']} | Qty: {tx['qty']} | Line Total: {line_total}")

print(f"\nOverall Dataset Transaction Value: {grand_total}")
```
